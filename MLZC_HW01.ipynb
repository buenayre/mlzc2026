```python
import pandas as pd
import numpy as np

#load csv
df = pd.read_csv('car_fuel_efficiency_2026.csv')

#test
print(df.head())

```

```python
#1 Records count
rcount = len(df)
print(rcount)
```

```python

#2 Unique fuels.
unique_fuel = df['fuel_type'].unique()
print(unique_fuel)
```

```python
#3 Columns with Null values
col_with_nulls = df.isnull().any().sum()
print(col_with_nulls)
```

```python
#4 max MPG
max_fuel = df['fuel_efficiency_mpg'].max()
print(max_fuel)
```

```python
#5 HP median
# Initial median
initial_med = df['horsepower'].median()
print(f"Initial Median: {initial_med}")

# Most frequent value
# In case the result are 2 values, [0] gets the first one.
mfv = df['horsepower'].mode()[0]
print(f"Most freq value: {mfv}")

# Filling missing values
df['horsepower'] = df['horsepower'].fillna(mfv)

# Calc median again
final_median = df['horsepower'].median()
print(f"Final Median (post-fillna): {final_median}")
```

```python
#6 Sum of weights

# 1. select asian cars
df_asia = df[df['origin'] == 'Asia']

# 2. needed columns
df_subset = df_asia[['vehicle_weight', 'model_year']]

# 3. first seven values
df_7 = df_subset.head(7)

# 4. numpy array (X)
X = df_7.to_numpy()

# 5. Matrix multiplication
XTX = X.T @ X  # np.dot(X.T, X) also works

# 6. Matrix inverse
XTX_inv = np.linalg.inv(XTX)

# 7. New vector
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Calc weight vector 'w'
w = XTX_inv @ X.T @ y

# 9. (w) Sum)
total_sum = w.sum()

print(f"Result w: {w}")
print(f"Final sum: {total_sum}")
```
